### Question 6: Median value of horsepower

1. Find the median value of the `horsepower` column in the dataset.
2. Next, calculate the most frequent value of the same `horsepower` column.
3. Use the `fillna` method to fill the missing values in the `horsepower` column with the most frequent value from the previous step.
4. Now, calculate the median value of `horsepower` once again.

Has it changed?
- Yes, it increased
- Yes, it decreased
- No


```python
# 1. Median value before imputation
median_before = df['horsepower'].median()
print(f"Median before imputation: {median_before}")

# 2. Most frequent value (mode)
mode_hp = df['horsepower'].mode()[0]
print(f"Most frequent value (mode): {mode_hp}")

# 3. Fill missing values with the mode
df_filled = df.copy()
df_filled['horsepower'] = df_filled['horsepower'].fillna(mode_hp)

# 4. Median value after imputation
median_after = df_filled['horsepower'].median()
print(f"Median after imputation: {median_after}")

# Comparison
if median_after > median_before:
    print("Has it changed?: Yes, it increased")
elif median_after < median_before:
    print("Has it changed?: Yes, it decreased")
else:
    print("Has it changed?: No")
```

**Answer:** `Yes, it decreased`

The median changed from **254.0** to **252.0** after filling missing values with the mode (252.0).


### Question 7: Sum of weights

1. Select all the cars from Asia
2. Select only columns `vehicle_weight` and `model_year`
3. Select the first 7 values
4. Get the underlying NumPy array. Let's call it `X`.
5. Compute matrix-matrix multiplication between the transpose of `X` and `X`. To get the transpose, use `X.T`. Let's call the result `XTX`.
6. Invert `XTX`.
7. Create an array `y` with values `[1100, 1300, 800, 900, 1000, 1100, 1200]`.
8. Multiply the inverse of `XTX` with the transpose of `X`, and then multiply the result by `y`. Call the result `w`.
9. What's the sum of all the elements of the result?
- 0.0369
- 0.369
- 3.69
- 36.9


```python
# 1. Select all the cars from Asia
# 2. Select only columns vehicle_weight and model_year
# 3. Select the first 7 values
asia_cars = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']].head(7)

# 4. Get the underlying NumPy array
X = asia_cars.to_numpy()
print("X:\n", X)

# 5. Compute matrix-matrix multiplication between X.T and X
XTX = X.T @ X
print("\nXTX:\n", XTX)

# 6. Invert XTX
XTX_inv = np.linalg.inv(XTX)
print("\nXTX_inv:\n", XTX_inv)

# 7. Create array y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
print("\ny:\n", y)

# 8. Multiply inverse of XTX with X.T, and then multiply by y
w = XTX_inv @ X.T @ y
print("\nw:\n", w)

# 9. Sum of all elements of w
sum_w = w.sum()
print(f"\nSum of weights: {sum_w:.4f} -> rounded to 3 decimal places: {sum_w:.3f}")
```

**Answer:** `0.369`

The computed sum of weights is approximately **0.3692**, which matches the option **`0.369`**.
